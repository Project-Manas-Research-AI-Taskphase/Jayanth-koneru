```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ts = pd.read_csv('Regression/Salary_Data.csv')
x_train = ts['YearsExperience'].values
y_train = ts['Salary'].values

```

```python
plt.scatter(x_train,y_train)
plt.xlabel('Years of Experience')
plt.ylabel('Salary of person')
plt.show()
```

```python
def cost_function(x, y, w, b):
  m = len(x)
  cost = 0

  for i in range(m):
    f = w * x[i] + b
    cost += (f - y[i])**2

  cost = (1 / (2 * m)) * cost

  return cost
```

```python
def gradient(x,y,w,b):
  m = len(x)
  dc_db = 0
  dc_dw = 0

  for i in range(m):
    f = w * x[i] + b
    dc_db += (f - y[i])
    dc_dw += (f - y[i]) * x[i]

  dc_db = 1/m * dc_db
  dc_dw = 1/m * dc_dw

  return dc_dw, dc_db



```

```python

def gradient_descent(x,y,alpha,iter):
  w = 0
  b = 0

  for i in range(iter):
    dc_db, dc_dw = gradient(x,y,w,b)

    w = w - alpha * dc_dw
    b = b - alpha * dc_db

    print(f"iter {i}: cost {cost_function(x, y, w, b)}")

  return w,b
```

```python
lr = 0.01
iter = 5

fw, fb = gradient_descent(x_train,y_train,lr,iter)

print (f"the final W = {fw:.4f} \n The final b = {fb:.4f}")
```
